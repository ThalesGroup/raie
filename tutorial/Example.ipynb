```python
from raie import fetch_database
from raie.window import windowing
from pathlib import Path
import pandas as pd
```

```python
# Define the base data folder
base = Path("data")

# Define subfolders
subfolders = ["fantasia", "gudb", "mit_arrhythmia", "mit_normal",'ludb']

# Create all directories
for sub in subfolders:
    (base / sub).mkdir(parents=True, exist_ok=True)
```

STEP 1: Download the databases to use them:
- mit_normal: https://physionet.org/content/nsrdb/1.0.0/
- fantasia: https://physionet.org/content/fantasia/1.0.0/
- mit_arrhythmia: https://alpha.physionet.org/content/mitdb/1.0.0/
- ludb: https://physionet.org/content/ludb/1.0.1/

STEP 2: Extract each database to its corresponding folder that was created in the previous cell.

For example, for Fantasia database, this would be the path for a data file relative to the root of this repository:
 `tutorial\data\fantasia\fantasia-database-1.0.0\f1o01.dat`


```python
# Fetch and format databases
fetch_database.download_and_format_gudb(output_path="data/gudb/")

# mit_normal: https://physionet.org/content/nsrdb/1.0.0/
fetch_database.download_and_format_mit_normal(database_path='data/mit_normal/', output_path='data/mit_normal/')

# fantasia: https://physionet.org/content/fantasia/1.0.0/
fetch_database.download_and_format_fantasia(database_path='data/fantasia/', output_path='data/fantasia/')

# mit_arrhythmia: https://alpha.physionet.org/content/mitdb/1.0.0/
fetch_database.download_and_format_mit_arrhythmia(database_path='data/mit_arrhythmia/', output_path='data/mit_arrhythmia/')

# ludb: https://physionet.org/content/ludb/1.0.1/
fetch_database.download_and_format_ludb(database_path='data/ludb/',output_path='data/ludb/')
```

```python
database = "./data/ludb" # Database path
dataFile = pd.read_csv(database+"/ECGs.csv") # Path to where the ECGs.csv file is stored
rpeaksFile = pd.read_csv(database+"/Rpeaks.csv") # Path to where the Rpeaks.csv file is stored
winlen = [30,100,150] # List of window lengths to try
for i in winlen:
    print("Window_length = "+str(i))
    outputFile = "./real_martinez_w"+str(i)+"_e9_fixing.csv" # Example of output path
    results_windowing = windowing(dataFile, rpeaksFile, window_length=i, step_size=1, edge_threshold=9, outputFile = outputFile, detection_method="martinez2004")
```
